```python
import csv, pandas as pd, matplotlib as plt, seaborn as sns
from IPython.display import display

#Leitura e extração dos dados utilizando csv.DictReader
with open("archive/Base Varejo.csv", encoding="utf-8") as arquivo:
    dados = list(csv.DictReader(arquivo, delimiter=";"))

# Transformação dos dados em dataframe utilizando o Pandas
df = pd.DataFrame(dados)

# Verificações iniciais
display(df.shape) # número de registros e colunas
display(df.columns) # nomes das colunas
display(df.dtypes) # tipos de dados
```
