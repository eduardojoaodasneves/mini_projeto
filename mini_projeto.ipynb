# Sprint 1 - Importacão das bibliotecas, do csv e primeiras análises.

```python
import csv, pandas as pd, matplotlib as plt, seaborn as sns
from IPython.display import display

#Leitura e extração dos dados utilizando csv.DictReader
with open("archive/Base Varejo.csv", encoding="utf-8") as arquivo:
    dados = list(csv.DictReader(arquivo, delimiter=";"))

# Transformação dos dados em dataframe utilizando o Pandas
df = pd.DataFrame(dados)

# Verificações iniciais
display(df.shape) # número de registros e colunas
display(df.columns) # nomes das colunas
display(df.dtypes) # tipos de dados
```

# Sprint 2 - Limpeza de texto, inteiros e decimais.

```python
display(df.dtypes)

# Conversão da coluna DATA
df["DATA"] = pd.to_datetime(df["DATA"], dayfirst=True)

# Conversão para inteiros
df["CO_ID"] = df["CO_ID"].apply(int)
df["CL_ID"] = df["CL_ID"].apply(int)
df["CL_EC"] = df["CL_EC"].apply(int)
df["CL_FHL"] = df["CL_FHL"].apply(int)
df["PR_ID"] = df["PR_ID"].apply(int)

# Limpeza das strings
df["PR_NOME"] = df["PR_NOME"].str.strip()
df["PR_CAT"] = df["PR_CAT"].str.strip()

display(df.dtypes)
```

 # Sprint 3 - Limpeza de Nulos e duplicatas.

```python

# Verificação de valores nulos
display(df.isna().sum())


# Verificação de duplicatas
display(df.duplicated().sum())


# Remoção das duplicatas
df = df.drop_duplicates()


# Verificação após a remoção das duplicatas
display(df.shape)
display(df.duplicated().sum())


# Verificação dos valores #N/D
display(df.isin(["#N/D"]).sum())


# Tratamento das categorias sem informação
if "#N/D" in df["PR_CAT"].values:
    df["PR_CAT"] = df["PR_CAT"].replace("#N/D", "Sem Categoria")
else:
    df["PR_CAT"] = df["PR_CAT"]


# Tratamento dos produtos sem informação
if "#N/D" in df["PR_NOME"].values:
    df["PR_NOME"] = df["PR_NOME"].replace("#N/D", "Sem Nome")
else:
    df["PR_NOME"] = df["PR_NOME"]


# Verificação dos valores #N/D após o tratamento
display(df.isin(["#N/D"]).sum())

display(df.isna().sum())


# Correção dos tipos de dados após as transformações
df["DATA"] = pd.to_datetime(df["DATA"], dayfirst=True)

df["CO_ID"] = df["CO_ID"].apply(int)
df["CL_ID"] = df["CL_ID"].apply(int)
df["CL_EC"] = df["CL_EC"].apply(int)
df["CL_FHL"] = df["CL_FHL"].apply(int)
df["PR_ID"] = df["PR_ID"].apply(int)


# Verificação final dos tipos
display(df.dtypes)


# Verificação final
display(df.shape)
display(df.duplicated().sum())
display(df.isna().sum())
display(df.isin(["#N/D"]).sum())
```
